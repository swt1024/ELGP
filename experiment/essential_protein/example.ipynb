```python
# Human
import pandas as pd

# =========================
# 1. Load input files
# =========================

lnc_mapping = pd.read_csv("../../data/LPI/human/lncRNA_mapping.csv")
lnc = pd.read_csv("../../data/LPI/human/lncRNA.csv")
ep = pd.read_csv("human_ess_protein_id.csv")
lpi = pd.read_csv("../../data/LPI/human/lpi.csv")

# Keep only the required columns
lnc_mapping = lnc_mapping[["lncRNA_id", "member_id"]]
lnc = lnc[["identifier", "gene_name", "gene_id"]]
ep = ep[["protein_id"]]
ep["protein_id"] = ep["protein_id"].astype(str).str.strip().str.replace(r"^p", "", regex=True)
lpi = lpi[["identifier", "protein"]]

# Remove duplicates
lnc_mapping = lnc_mapping.drop_duplicates()
lnc = lnc.drop_duplicates()
ep = ep.drop_duplicates()
lpi = lpi.drop_duplicates()

# =========================
# 2. Process each tissue
# =========================

for t in ['heart', 'lung', 'stomach']:
    # Read essential merged lncRNA IDs
    ess = pd.read_csv(
        f"../ess_number/filtered/human/BC_top40pct_human_{t}_esslnc.csv",
        header=None
    )
    ess.columns = ["lncRNA_id"]

    # Step 1: keep only essential merged lncRNAs in mapping table
    ess_mapping = lnc_mapping[lnc_mapping["lncRNA_id"].isin(ess["lncRNA_id"])].copy()

    # Step 2: attach original lncRNA annotation using member_id -> identifier
    ess_annot = ess_mapping.merge(
        lnc,
        left_on="member_id",
        right_on="identifier",
        how="left"
    )

    # Step 3: attach lncRNA-protein interactions using identifier
    ess_lpi = ess_annot.merge(
        lpi,
        on="identifier",
        how="inner"
    )

    # Step 4: keep only essential proteins
    ess_lpi = ess_lpi[ess_lpi["protein"].isin(ep["protein_id"])].copy()

    # Step 5: keep required columns
    out = ess_lpi[["lncRNA_id", "gene_name", "gene_id", "protein"]].drop_duplicates()

    # Save output
    out.to_csv(f"elp/human_{t}_essential_lncRNA_essential_protein_pairs.csv", index=False)

    print(f"{t}: {len(out)} pairs saved.")
```
